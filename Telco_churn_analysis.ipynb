## Load the data

Change the file name or path if your CSV is somewhere else.

```python
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")
print(df.shape)
df.head()
```

```python
df.info()
```

## Clean it and check the target

`Churn` becomes 1 (left) / 0 (stayed). The classes are unbalanced, so accuracy alone is misleading.

```python
df = clean_data(df)
print(df['Churn'].value_counts())
print('Churn rate:', round(df['Churn'].mean(), 3))
```

## Churn rate by category

```python
for col in ['Contract', 'InternetService', 'PaymentMethod']:
    print(df.groupby(col)['Churn'].mean().round(3).sort_values(ascending=False), '\n')
```

## Numeric columns vs churn

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, col in zip(axes, ['tenure', 'MonthlyCharges', 'TotalCharges']):
    df.boxplot(column=col, by='Churn', ax=ax)
plt.suptitle('')
plt.show()
```

## My observations

- Churners are the minority class, so the data is imbalanced and accuracy alone would be misleading.
- Customers on month-to-month contracts churn much more than customers on one-year or two-year contracts. The longer the contract, the lower the churn.
- Churn is highest among customers with short tenure and falls steadily as tenure grows, so new customers are the most at risk.
- Customers with higher monthly charges tend to churn more than customers with lower bills.
- Fiber optic customers churn more than DSL customers, and customers without internet service churn the least.
- Paying by electronic check is linked to higher churn than automatic payment methods (bank transfer or credit card).
- Customers without online security or tech support churn more than customers who have these add-ons.
- Gender has little effect on churn, and customers with a partner or dependents churn somewhat less.
